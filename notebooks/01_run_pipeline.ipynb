## Configuración

* `configs/ci.yaml` — datos sintéticos + modo rápido (segundos/minutos; ejecuta antes `eduoutcomes make-synthetic`).
* `configs/quick.yaml` — datos reales, rejillas mínimas (**no reportable**).
* `configs/default.yaml` — diseño preespecificado del manuscrito.

```python
cfg = load_config("configs/quick.yaml", overrides=["show_figures=True"])
print(cfg.output_dir, cfg.quick)
```

```python
# Ejecuta hasta la comparación estadística (PI1). Usa until=None para todo el pipeline.
ctx = run(cfg, until="rq1")
```
